```python
import pandas as pd
from PIL import Image
```

```python
data_file = "../data/uniqlo_data.csv"
df = pd.read_csv(data_file)
```

```python
df.head()
```

```python
df.columns,len(df.columns)
```

```python
df.iloc[0]["Color_options"]
```

```python
cols = ['color1_img1', 'color2_img1', 'color3_img1', 'color4_img1',
        'color5_img1', 'color6_img1', 'color7_img1', 'img2', 'img3', 'img4',
        'img5']
```

```python
len(df)
```

```python
data = []
for i in range(len(df)):
    for col in cols:
        t = df.iloc[i][col]
        if isinstance(t,str) and t.startswith("https:"):
            data.append({
                'id': df.iloc[i]['id'],
                'price': df.iloc[i]['price'],
                'item_url': df.iloc[i]['item_url'],
                'Item_name': df.iloc[i]['Item_name'],
                'Item_type': df.iloc[i]['Item_type'],
                'materials': df.iloc[i]['materials'],
                'img_url': t
            })
```

```python
len(data)
```

```python
df_processed = pd.DataFrame(data)
```

```python
df_processed.head()
```

```python
df_processed.to_csv("../data/uniqlo_processed.csv",index=False)
```

```python
## removing bad images
df = pd.read_csv("../data/uniqlo_processed.csv")
```

```python
def is_good(x):
    try:
        Image.open(requests.get(x["img_url"], stream=True).raw)
        return True
    except:
        return False
    return True
```

```python
df["is_good"] = df.apply(lambda x: is_good(x),axis=1)
```

```python
df.to_csv("../data/uniqlo_processed.csv",index=False)
```
