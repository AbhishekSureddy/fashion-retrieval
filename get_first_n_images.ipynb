```python
from dataloader.dataloader import UniqloDataset
from tqdm import tqdm
import json
```

```python
uniqlo_path = "./data/uniqlo_processed.csv"
```

```python
dl = UniqloDataset(uniqlo_path)
```

```python
dl[0]
```

```python
save_dir = "./sample_images/"
```

```python
def process_dl_item(dl_item,img_file):
    it_ = {}
    keys = ['idx','id','price', 'item_url', 'Item_name', 'Item_type', 'materials', 'img_url']
    for k in keys:
        it_[k] = dl_item[k]
    it_["image_file"] = img_file
    return it_
```

```python
n = 500
```

```python
descriptions = []
for i in tqdm(range(n)):
    img = dl[i]['image']
    img_file = f"{save_dir}/{i}.png"
    img.save(img_file)
    descr = process_dl_item(dl[i], img_file)
    descriptions.append(descr)
```

```python
with open('sample_images.json','w') as f:
    json.dump(descriptions,f)
```
