```python
import matplotlib.pyplot as plt
import math
from PIL import Image

def plot_images_grid(images, figsize=(10, 10)):
    """
    Display images in a grid.
    
    Parameters:
    - images: list of images (as PIL Images or numpy arrays)
    - figsize: tuple specifying the figure size
    """
    # Calculate the number of rows and columns
    n = math.ceil(math.sqrt(len(images)))  # Number of columns
    m = math.ceil(len(images) / n)         # Number of rows

    # Create the figure with specified figsize
    fig, axs = plt.subplots(m, n, figsize=figsize)
    
    # Flatten the axis array for easy iteration
    axs = axs.flatten()

    for i, img in enumerate(images):
        axs[i].imshow(img)
        axs[i].axis('off')  # Hide axes for better visual

    # Turn off any unused subplots
    for j in range(i + 1, len(axs)):
        axs[j].axis('off')

    plt.tight_layout()
    plt.show()

def visualize_ids(ims_list):
    # Load images into a list
    images = [Image.open(f"{img_dir}/{i}.png") for i in ims_list]  # Replace with your image paths
    plot_images_grid(images)
```

```python
img_dir = "./sample_images/"
# use this to give list of images to see
# you can give like [1,2,3,4]; here, 1,2,3,4,... are idx in the sample_images.json
ims_list = [1,2,3,4] ## you can change this
```

```python
visualize_ids(ims_list)
```

```python
# If you want to generate images from a start idx to end idx
# It includes start index, excludes end index
start = 0
end = 50
ims_list = list(range(start,end))
visualize_ids(ims_list)
```
